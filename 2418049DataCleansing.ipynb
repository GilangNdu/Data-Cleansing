**2. Import library**

```python
import pandas as pd
import numpy as np
```

**3. Membaca dataset**

```python
df = pd.read_excel("Dataset_Mahasiswa_Kehadiran_Aktivitas_IPK.xlsx")
print(df.shape)
df.head()
```

**4. Melihat ukuran dataset**

```python
print("Jumlah baris :", df.shape[0])
print("Jumlah kolom :", df.shape[1])
```

**5. Melihat informasi dataset**

```python
df.info()
```

**6. Cek missing value**

```python
missing = df.isnull().sum()

print("Jumlah missing value setiap kolom:")
print(missing)
```

**7. Cek data duplikat**

```python
print("Jumlah data duplikat:", df.duplicated().sum())
```

**8. Bersihkan spasi pada data teks**

```python
kolom_teks = df.select_dtypes(include="object").columns

for kolom in kolom_teks:
    df[kolom] = df[kolom].str.strip()
```

**9. Cek kategori yang tidak konsisten**

```python
for kolom in ["Jenis Kelamin", "Status Menikah", "Status Akademik"]:
    print(f"\n{kolom}:")
    print(df[kolom].value_counts())
```

**10. Cek nilai unik data kategorikal**

```python
print("Jenis Kelamin:")
print(df["Jenis Kelamin"].unique())

print("\nStatus Menikah:")
print(df["Status Menikah"].unique())

print("\nStatus Akademik:")
print(df["Status Akademik"].unique())
```

**11. Cek rentang nilai numerik**

```python
kolom_numerik = [
    "Umur",
    "Kehadiran (%)",
    "Partisipasi Diskusi (skor)",
    "Nilai Tugas (rata-rata)",
    "Aktivitas E-Learning (skor)",
    "IPK"
]

for kolom in kolom_numerik:
    df[kolom] = pd.to_numeric(df[kolom], errors="coerce")

print("Missing value setelah konversi tipe data:")
print(df[kolom_numerik].isnull().sum())

print("\nStatistik deskriptif:")
print(df[kolom_numerik].describe())
```

**12. Validasi nilai berdasarkan logika data**

```python
#13. Validasi nilai berdasarkan logika data
print("Kehadiran di luar 0-100:")
print(df[(df["Kehadiran (%)"] < 0) | (df["Kehadiran (%)"] > 100)])

#Partisipasi diskusi
print("Partisipasi di luar 0-100:")
print(df[
    (df["Partisipasi Diskusi (skor)"] < 0) |
    (df["Partisipasi Diskusi (skor)"] > 100)
])

#Nilai tugas
print("Nilai tugas di luar 0-100:")
print(df[
    (df["Nilai Tugas (rata-rata)"] < 0) |
    (df["Nilai Tugas (rata-rata)"] > 100)
])

#Aktivitas E-Learning
print("Aktivitas E-Learning di luar 0-100:")
print(df[
    (df["Aktivitas E-Learning (skor)"] < 0) |
    (df["Aktivitas E-Learning (skor)"] > 100)
])


#IPK
print("IPK di luar 0-4:")
print(df[
    (df["IPK"] < 0) |
    (df["IPK"] > 4)
])

```

**13. Cek outlier**

```python
for kolom in kolom_numerik:
    Q1 = df[kolom].quantile(0.25)
    Q3 = df[kolom].quantile(0.75)
    IQR = Q3 - Q1

    batas_bawah = Q1 - 1.5 * IQR
    batas_atas = Q3 + 1.5 * IQR

    jumlah_outlier = (
        (df[kolom] < batas_bawah) |
        (df[kolom] > batas_atas)
    ).sum()

    print(f"{kolom}: {jumlah_outlier} outlier")
```

**14. Cek data setelah cleaning**

```python
print("Jumlah data akhir:", df.shape)

print("\nMissing value:")
print(df.isnull().sum())

print("\nDuplikat:", df.duplicated().sum())
```

**15. Simpan dataset hasil cleaning**

```python
df.to_excel("Dataset_Mahasiswa_Clean.xlsx", index=False)
print("Dataset berhasil disimpan.")



from google.colab import files
files.download("Dataset_Mahasiswa_Clean.xlsx")
```
